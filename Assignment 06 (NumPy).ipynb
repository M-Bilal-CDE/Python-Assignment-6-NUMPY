1) Create a 1D NumPy array of integers from 10 to 50 (inclusive, step 5). Print the array, its shape, dtype and number of dimensions.

```python
import numpy as np

arr = np.linspace(10, 50, 9, dtype=int)

print("Array:", arr)
print("Shape:", arr.shape)
print("Dtype:", arr.dtype)
print("Number of dimensions:", arr.ndim)
```

2) Create a 3x4 array of zeros, a 3x4 array of ones, and a 4x4 identity matrix. Then create an array of 8 evenly spaced numbers between 0 and 1 using linspace.

```python
zeros_arr = np.zeros((3, 4))
print("Zeros Array (3x4):\n", zeros_arr)

ones_arr = np.ones((3, 4))
print("\nOnes Array (3x4):\n", ones_arr)

identity_arr = np.eye(4)
print("\nIdentity Matrix (4x4):\n", identity_arr)

linspace_arr = np.linspace(0, 1, 8)
print("\nLinspace Array (8 numbers):\n", linspace_arr)
```

3) Create a 1D array of numbers from 1 to 24 and reshape it into a 2D array of shape (4, 6). Then reshape it into a 3D array of shape (2, 3, 4) and print the shape of each.

```python
arr_1d = np.arange(1, 25)
print("1D Array:", arr_1d)

arr_2d = arr_1d.reshape(4, 6)
print("\n2D Array (4, 6):\n", arr_2d)
print("Shape of 2D array:", arr_2d.shape)

arr_3d = arr_1d.reshape(2, 3, 4)
print("\n3D Array (2, 3, 4):\n", arr_3d)
print("Shape of 3D array:", arr_3d.shape)
```

4) Given arr = np.array([12, 45, 7, 23, 56, 89, 34, 21]), print: the first 3 elements, the last 3 elements, every alternate element, and the array in reverse order.

```python
arr = np.array([12, 45, 7, 23, 56, 89, 34, 21])

print("First 3 elements:", arr[:3])
print("Last 3 elements:", arr[-3:])
print("Every alternate element:", arr[::2])
print("Reverse order:", arr[::-1])
```

5) Create a 2D array of shape (5, 5) with values 1 to 25. Extract (a) the second row, (b) the third column, (c) the 3x3 sub-matrix from the center.

```python
import numpy as np
arr = np.arange(1, 26).reshape(5, 5)
print("Original Array:\n", arr)
second_row = arr[1]
print("\n(a) Second row:", second_row) 
third_col = arr[:, 2]
print("(b) Third column:", third_col)
```

6) Given a = np.array([10, 20, 30, 40]) and b = np.array([1, 2, 3, 4]), perform addition, subtraction, multiplication, division and power (a ** 2). Also compute the dot product of a and b.

```python
a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

print("Addition (a + b):", a + b)
print("Subtraction (a - b):", a - b)
print("Multiplication (a * b):", a * b)

print("Division (a / b):", a / b)

print("Power (a ** 2):", a ** 2)

dot_prod = np.dot(a, b)
print("Dot Product:", dot_prod)
```

7) Given arr = np.array([[4, 8, 2], [9, 1, 7], [5, 6, 3]]), find the sum, mean, min, max and standard deviation of the whole array, and then the sum along rows (axis=1) and along columns (axis=0).

```python
import numpy as np

arr = np.array([[4, 8, 2], 
                [9, 1, 7], 
                [5, 6, 3]])

print("Sum of whole array:", np.sum(arr))
print("Mean of whole array:", np.mean(arr))
print("Min of whole array:", np.min(arr))
print("Max of whole array:", np.max(arr))
print("Standard Deviation:", np.std(arr))
print("\nSum along rows (axis=1):", np.sum(arr, axis=1))
print("Sum along columns (axis=0):", np.sum(arr, axis=0))
```

8) Create an array of 20 random integers between 1 and 100. Using boolean masking, print (a) all elements greater than 50, (b) all even elements, (c) count of elements divisible by 5.

```python
import numpy as np

arr = np.random.randint(1, 101, size=20)
print("Original Array:\n", arr)
greater_50 = arr[arr > 50]
print("\n(a) Elements greater than 50:", greater_50)
even_arr = arr[arr % 2 == 0]
print("(b) Even elements:", even_arr)
div_by_5_count = np.sum(arr % 5 == 0)
print("(c) Count of elements divisible by 5:", div_by_5_count)
```

9) Use np.where to replace all negative values in np.array([5, -3, 8, -1, 0, -7, 4]) with 0. Also find the index of the maximum and minimum value using argmax and argmin.

```python
import numpy as np

arr = np.array([5, -3, 8, -1, 0, -7, 4])
print("Original Array:", arr)
replaced_arr = np.where(arr < 0, 0, arr)
print("Array after replacing negatives with 0:", replaced_arr)
max_index = np.argmax(arr)
min_index = np.argmin(arr)

print("Index of maximum value:", max_index)
print("Index of minimum value:", min_index)
```

10) Create two 2x3 arrays and join them vertically (vstack) and horizontally (hstack). Then sort the array np.array([[9, 3, 6], [2, 8, 1]]) along each row and flatten the result into 1D.

```python
import numpy as np
arr1 = np.array([[1, 2, 3], 
                 [4, 5, 6]])

arr2 = np.array([[7, 8, 9], 
                 [10, 11, 12]])
v_stacked = np.vstack((arr1, arr2))
print("Vertically stacked (vstack):\n", v_stacked)
h_stacked = np.hstack((arr1, arr2))
print("\nHorizontally stacked (hstack):\n", h_stacked)
arr = np.array([[9, 3, 6], 
                [2, 8, 1]])
sorted_arr = np.sort(arr, axis=1)
flattened_arr = sorted_arr.flatten()

print("\nSorted along rows and flattened (1D):\n", flattened_arr)
```
